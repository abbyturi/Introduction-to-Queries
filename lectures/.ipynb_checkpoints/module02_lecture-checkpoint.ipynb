# Part 1: The Complete SELECT Framework


## What you already know

A Module 1 query may have looked like this:

```sql
SELECT title, rental_rate
FROM film
WHERE rating = 'PG'
ORDER BY rental_rate DESC
LIMIT 10;
```

You can already explain four actions:

- choose output columns
- identify the source table
- remove unwanted rows
- sort and trim the result

Module 2 asks a deeper question: **In what order does PostgreSQL logically perform those actions?**


## The six-clause framework

A complete analytical `SELECT` statement can use these six major clauses in this writing order:

```sql
SELECT   <columns or expressions>
FROM     <table or source>
WHERE    <row conditions>
GROUP BY <grouping columns>
HAVING   <group conditions>
ORDER BY <sort expressions>;
```

`LIMIT` may follow `ORDER BY` to cap the number of rows returned.


## Writing order is not execution order

The order you type the clauses is not the logical order PostgreSQL uses to process them.

```text
WRITING ORDER              LOGICAL EXECUTION ORDER
-------------              -----------------------
SELECT                      1. FROM
FROM                        2. WHERE
WHERE                       3. GROUP BY
GROUP BY                    4. HAVING
HAVING                      5. SELECT
ORDER BY                    6. ORDER BY
LIMIT                       7. LIMIT
```

This difference explains several common SQL errors.


## Why `FROM` logically comes first

PostgreSQL must first know **which rows exist for the query to work with**.

```sql
FROM film
```

establishes the source relation. Only after the source exists can PostgreSQL evaluate row-level conditions.

### Think about it

If `WHERE` logically ran before `FROM`, what rows would it filter?


## Step 2: `WHERE`

`WHERE` removes individual rows before grouping and before the final output columns are defined.

```sql
SELECT title, rating
FROM film
WHERE rating = 'PG';
```

At the `WHERE` step, PostgreSQL knows columns from `film`, but it does **not** yet know aliases that will be created later in `SELECT`.


## Step 3: `GROUP BY`

`GROUP BY` organizes surviving rows into groups.

```sql
GROUP BY customer_id
```

You will study grouping later. For Module 2, remember only this sequence:

```text
rows from FROM
      ↓
rows surviving WHERE
      ↓
groups formed by GROUP BY
```


## Step 4: `HAVING`

`HAVING` filters groups after grouping has occurred.

```sql
SELECT customer_id, COUNT(*) AS rental_count
FROM rental
GROUP BY customer_id
HAVING COUNT(*) > 30;
```

A useful temporary memory aid is:

> `WHERE` filters rows; `HAVING` filters groups.


## Step 5: `SELECT`

Only now does PostgreSQL determine the requested output expressions and define their aliases.

```sql
SELECT title,
       length,
       length / 60.0 AS duration_hours
FROM film;
```

The alias `duration_hours` comes into existence at the `SELECT` step.


## Step 6: `ORDER BY`

Because `ORDER BY` logically follows `SELECT`, it can use a `SELECT` alias.

```sql
SELECT title,
       length / 60.0 AS duration_hours
FROM film
ORDER BY duration_hours DESC;
```

This is valid PostgreSQL.


## Step 7: `LIMIT`

`LIMIT` trims the already-formed result.

```sql
SELECT title, replacement_cost
FROM film
ORDER BY replacement_cost DESC
LIMIT 10;
```

Without a meaningful `ORDER BY`, the phrase “top 10” has no analytical meaning.

> **Remember**
>
> `LIMIT 10` means “return at most 10 rows.” It does not mean “return the 10 highest, lowest, newest, or oldest rows” unless `ORDER BY` defines that ordering.


# Part 2: Execution Order Errors


## Error 1: using a SELECT alias in WHERE

Consider this query:

```sql
SELECT title,
       length / 60.0 AS duration_hours
FROM film
WHERE duration_hours > 2;
```

### Predict before running

Will PostgreSQL accept `duration_hours` in `WHERE`?


## Why the alias query fails

Logical order:

```text
1. FROM film
2. WHERE duration_hours > 2   ← alias does not exist yet
...
5. SELECT length / 60.0 AS duration_hours
```

`WHERE` runs before `SELECT`, so the alias is unavailable.


## Correct the alias error

Use the underlying column or expression in `WHERE`:

```sql
SELECT title,
       length,
       length / 60.0 AS duration_hours
FROM film
WHERE length > 120
ORDER BY duration_hours DESC;
```

Notice that `ORDER BY` **can** use the alias because it logically runs after `SELECT`.


## Quick check: alias availability

For the alias `weekly_cost` created below, decide whether each use is valid.

```sql
SELECT title,
       rental_rate * 7 AS weekly_cost
FROM film
WHERE rental_rate < 3
ORDER BY weekly_cost;
```

1. `WHERE weekly_cost < 10`
2. `ORDER BY weekly_cost`

### Answer

1. **Invalid** — `WHERE` runs before `SELECT`.
2. **Valid** — `ORDER BY` runs after `SELECT`.


## Error 2: aggregate function in WHERE

This query is also incorrect:

```sql
SELECT customer_id,
       COUNT(*) AS rental_count
FROM rental
WHERE COUNT(*) > 30
GROUP BY customer_id;
```

At the `WHERE` step, the groups do not exist yet.


## Correct aggregate filtering

Use `HAVING` for a condition on grouped results:

```sql
SELECT customer_id,
       COUNT(*) AS rental_count
FROM rental
GROUP BY customer_id
HAVING COUNT(*) > 30
ORDER BY rental_count DESC;
```

> **For now**
>
> You do not need to master `COUNT`, `GROUP BY`, or `HAVING` yet. You do need to recognize that an aggregate condition cannot be evaluated in `WHERE` before grouping occurs.


## Try it: trace all steps

Run this query and trace it in logical order:

```sql
SELECT customer_id,
       COUNT(*)         AS num_rentals,
       MIN(rental_date) AS first_rental,
       MAX(rental_date) AS most_recent
FROM rental
WHERE rental_date >= '2005-07-01'
GROUP BY customer_id
HAVING COUNT(*) > 20
ORDER BY num_rentals DESC
LIMIT 10;
```

### Trace questions

1. Which rows enter from `rental`?
2. Which rows survive `WHERE`?
3. What creates the customer groups?
4. Which groups survive `HAVING`?
5. When is `num_rentals` created?
6. Why can `ORDER BY` use `num_rentals`?
7. What does `LIMIT 10` do last?


# Part 3: Comparison Operators


## Six basic comparison operators

| Operator | Meaning | Example |
|---|---|---|
| `=` | equal to | `rating = 'PG'` |
| `<>` or `!=` | not equal to | `rating <> 'R'` |
| `<` | less than | `length < 60` |
| `>` | greater than | `rental_rate > 2.99` |
| `<=` | less than or equal to | `replacement_cost <= 19.99` |
| `>=` | greater than or equal to | `rental_duration >= 5` |

These operators can be used with appropriate numeric, text, and date values.


## Be precise about the boundary

Compare:

```sql
WHERE replacement_cost > 20.00
```

with:

```sql
WHERE replacement_cost >= 20.00
```

The second condition includes exactly `20.00`; the first does not.

### Self-check

If a film costs exactly `20.00`, which condition includes it?


## Try it: numeric comparison

```sql
SELECT title, rental_rate, replacement_cost
FROM film
WHERE replacement_cost > 20.00
ORDER BY replacement_cost DESC, title;
```

Then change `>` to `>=`.

**Observe:** Does the row count change? If so, what does that tell you about values at the boundary?


# Part 4: Combining Conditions


## `AND`: all conditions must be true

```sql
SELECT title, rating, rental_rate
FROM film
WHERE rating = 'PG'
  AND rental_rate <= 2.99;
```

A row survives only when **both** conditions are true.


## `OR`: at least one condition must be true

```sql
SELECT title, rating
FROM film
WHERE rating = 'G'
   OR rating = 'PG-13';
```

A row survives if either condition is true.


## `NOT`: invert a condition

```sql
SELECT title, rating
FROM film
WHERE NOT rating = 'NC-17';
```

This asks for rows where the condition `rating = 'NC-17'` is not true.


## Operator precedence matters

PostgreSQL evaluates `AND` before `OR`.

Consider:

```sql
WHERE rating = 'G'
   OR rating = 'PG-13'
  AND rental_duration >= 6
```

PostgreSQL interprets the logic as:

```sql
WHERE rating = 'G'
   OR (rating = 'PG-13' AND rental_duration >= 6)
```

That may not be what the analyst intended.


## Parentheses communicate intent

If the business question is:

> Films rated G or PG-13, and in either case the rental duration must be at least 6 days

write:

```sql
WHERE (rating = 'G' OR rating = 'PG-13')
  AND rental_duration >= 6
```

> **Professional habit**
>
> When `AND` and `OR` appear in the same condition, use parentheses to make the intended logic explicit—even when you know the precedence rules.


## Try it: precedence experiment

Run both versions and compare row counts.

### Version A

```sql
SELECT title, rating, rental_duration
FROM film
WHERE (rating = 'G' OR rating = 'PG-13')
  AND rental_duration >= 6;
```

### Version B

```sql
SELECT title, rating, rental_duration
FROM film
WHERE rating = 'G' OR rating = 'PG-13'
  AND rental_duration >= 6;
```

### Explain the difference

Find one row returned by Version B that helps demonstrate why the two result sets differ.


# Part 5: Filtering Ranges with BETWEEN


## `BETWEEN` expresses an inclusive range

```sql
SELECT title, length
FROM film
WHERE length BETWEEN 45 AND 90
ORDER BY length;
```

`BETWEEN 45 AND 90` includes both `45` and `90`.


## Equivalent boundary logic

This:

```sql
WHERE length BETWEEN 45 AND 90
```

is equivalent to:

```sql
WHERE length >= 45
  AND length <= 90
```

Use the form that communicates the condition most clearly.


## The lower bound comes first

```sql
WHERE length BETWEEN 90 AND 45
```

is not a request for values from 45 through 90. The lower expression and upper expression have meaning.

### Quick check

Write a `BETWEEN` condition for replacement costs from `14.99` through `19.99`, including both values.


## `BETWEEN` with dates and timestamps

Date-like values are written as literals in single quotes:

```sql
SELECT rental_id, rental_date
FROM rental
WHERE rental_date BETWEEN '2005-06-14' AND '2005-06-16';
```

But `rental_date` is a timestamp. The upper literal may represent midnight at the beginning of June 16.


## Safer full-day timestamp ranges

For complete days, a half-open interval is often clearer:

```sql
SELECT rental_id, rental_date
FROM rental
WHERE rental_date >= '2005-06-14'
  AND rental_date <  '2005-06-17';
```

This includes all times on June 14, 15, and 16 while excluding June 17.

> **Timestamp boundary trap**
>
> For timestamp columns, think about the time portion. A date-looking upper bound can unintentionally exclude records later on that date.


# Part 6: Filtering Lists with IN


## `IN` replaces repeated OR conditions

Instead of:

```sql
WHERE rating = 'G'
   OR rating = 'PG'
   OR rating = 'PG-13'
```

write:

```sql
WHERE rating IN ('G', 'PG', 'PG-13')
```

Both express membership in a set of allowed values.


## Why `IN` is useful

`IN` is often:

- shorter
- easier to scan
- easier to modify
- less likely to accumulate repeated-condition mistakes

It works with compatible text, numeric, and date values.


## `NOT IN`

To exclude listed values:

```sql
SELECT title, rating
FROM film
WHERE rating NOT IN ('R', 'NC-17');
```

Read this as: “keep rows whose rating is not one of these values.”


## Try it: rewrite OR as IN

Start with:

```sql
SELECT title, rating
FROM film
WHERE rating = 'G'
   OR rating = 'PG'
   OR rating = 'PG-13';
```

Rewrite only the `WHERE` clause using `IN`.

Then compare the row counts.

### Reflection

Which version would be easier to maintain if the allowed list grew from 3 ratings to 8 values?


# Part 7: Pattern Matching


## Exact comparison is not always enough

This condition requires an exact text value:

```sql
WHERE title = 'ACADEMY DINOSAUR'
```

But analysts often need questions such as:

- Which titles start with `ACADEMY`?
- Which titles contain `LOVE` anywhere?
- Which names have exactly four characters?

For these questions, use pattern matching.


## `LIKE` and its two wildcards

| Wildcard | Meaning | Example pattern |
|---|---|---|
| `%` | zero or more characters | `'A%'` |
| `_` | exactly one character | `'B_ll'` |

PostgreSQL's `LIKE` matches the supplied pattern against the string, and the pattern normally describes the whole string.


## Starts with

```sql
SELECT title
FROM film
WHERE title LIKE 'ACADEMY%';
```

`%` allows any sequence of characters after `ACADEMY`.


## Contains

```sql
SELECT title
FROM film
WHERE title LIKE '%LOVE%';
```

The `%` on both sides allows characters before and after `LOVE`.


## Ends with

```sql
SELECT title
FROM film
WHERE title LIKE '%ING';
```

The pattern requires the text to end with `ING`.


## Exactly one character with `_`

```sql
SELECT first_name
FROM actor
WHERE first_name LIKE '____';
```

Four underscores mean exactly four characters.

### Predict

Would `AL` match? Would `ANNA` match? Would `JAMES` match?


## `LIKE` versus `ILIKE`

In PostgreSQL, `LIKE` is case-sensitive. `ILIKE` is a PostgreSQL extension for case-insensitive matching.

```sql
SELECT title
FROM film
WHERE title ILIKE '%love%';
```

This can match text regardless of capitalization according to the active locale.


## Pattern matching: choose the simplest tool

For Module 2:

- use `LIKE` for simple case-sensitive wildcard patterns
- use `ILIKE` for simple case-insensitive wildcard patterns
- recognize that PostgreSQL also supports more advanced pattern tools, but they are outside this module's required scope

> **From the PostgreSQL documentation**
>
> PostgreSQL documents three broad pattern-matching approaches: `LIKE`, `SIMILAR TO`, and POSIX regular expressions. This module concentrates on `LIKE`/`ILIKE` because they directly support the filtering skills needed for Lab 1.


## Try it: three pattern questions

Write and run three queries:

1. film titles beginning with `AGENT`
2. actor first names exactly four characters long
3. film titles containing `IRON` anywhere

### Check your wildcard placement

- starts with → text followed by `%`
- contains → `%` on both sides
- exact character count → one `_` per character


# Part 8: NULL Is Not an Ordinary Value


## What does NULL mean?

`NULL` represents an unknown, missing, or absent value.

It is **not**:

- zero
- an empty string
- the word “NULL”
- equal to another `NULL`

The meaning of missingness depends on the data context.


## A NULL can be meaningful

In a rental table, a `NULL` `return_date` may mean the film has not yet been returned.

```sql
SELECT rental_id, customer_id, rental_date
FROM rental
WHERE return_date IS NULL;
```

That is not necessarily a data-quality error. It may represent a real business state.


## Use `IS NULL`

Correct:

```sql
WHERE return_date IS NULL
```

Incorrect:

```sql
WHERE return_date = NULL
```

The second condition does not test missingness correctly.


## Use `IS NOT NULL`

```sql
SELECT rental_id, customer_id, return_date
FROM rental
WHERE return_date IS NOT NULL;
```

This keeps rows where a return date is present.


## Why `= NULL` does not work

SQL uses three-valued logic: a comparison involving an unknown value is generally **unknown**, not true.

So:

```sql
return_date = NULL
```

does not become true for missing values.

Use the dedicated predicates `IS NULL` and `IS NOT NULL`.


## Try it: profile missing values

```sql
SELECT COUNT(*) AS never_returned
FROM rental
WHERE return_date IS NULL;
```

Record the count from your Pagila instance.

### Why this matters

Before analyzing a field, ask whether missing values exist and what they mean. Missingness can affect filters, calculations, and interpretation.


## `NOT IN` and NULL: a caution

If the column being tested can contain `NULL`, do not assume `NOT IN (...)` automatically includes those missing rows.

For example:

```sql
WHERE rating NOT IN ('R', 'NC-17')
  AND rating IS NOT NULL
```

makes the intended treatment of missing ratings explicit.

> **Analytical habit**
>
> Whenever you use a negative filter such as `<>`, `NOT IN`, or `NOT LIKE`, ask: **What should happen to NULL rows?**


# Part 9: Shaping Query Output


## Filtering versus shaping

`WHERE` controls **which rows enter the result**.

`SELECT` controls **what the result looks like**.

Output shaping includes:

- choosing columns
- computing expressions
- assigning aliases
- removing duplicate output rows
- replacing `NULL` for presentation or calculation


## Expressions in SELECT

`SELECT` can calculate a value rather than simply display a stored column.

```sql
SELECT title,
       rental_rate,
       rental_rate * 7 AS weekly_cost
FROM film
ORDER BY weekly_cost DESC;
```

`weekly_cost` is computed for the result; it is not a new stored column in `film`.


## Aliases with AS

An alias gives an output expression a readable name.

```sql
SELECT customer_id,
       amount,
       amount * 1.10 AS amount_with_surcharge
FROM payment;
```

The alias helps the result communicate meaning.


## Quoted aliases

A simple lowercase alias is usually easiest:

```sql
AS amount_with_surcharge
```

If an alias contains spaces or intentionally preserved capitalization, double quotes are required:

```sql
AS "Amount With Surcharge"
```

For reusable analytical SQL, simple snake_case aliases are often easier to reference.


## String expressions

PostgreSQL uses `||` for string concatenation:

```sql
SELECT customer_id,
       first_name || ' ' || last_name AS full_name
FROM customer
ORDER BY last_name;
```

The literal `' '` inserts a space between the names.


## Try it: build an expression

Return:

- `customer_id`
- original `amount` as `payment_amount`
- `amount * 1.10` as `amount_with_surcharge`

Then sort by the surcharge amount descending and return 15 rows.

### Before running

Which alias can safely appear in `ORDER BY`? Why?


# Part 10: DISTINCT


## What DISTINCT does

`DISTINCT` removes duplicate rows from the selected output.

```sql
SELECT DISTINCT rating
FROM film
ORDER BY rating;
```

This answers a profiling question: **What unique rating values occur in the table?**


## DISTINCT applies to the selected combination

This query:

```sql
SELECT DISTINCT rating, rental_rate
FROM film;
```

returns unique **pairs** of `(rating, rental_rate)`.

It does not mean “one row per rating.”


## Try it: compare two DISTINCT queries

Run:

```sql
SELECT DISTINCT rating
FROM film
ORDER BY rating;
```

Then:

```sql
SELECT DISTINCT rating, rental_rate
FROM film
ORDER BY rating, rental_rate;
```

### Explain

Why does adding `rental_rate` change the number of distinct rows?


## DISTINCT is not a repair tool

`DISTINCT` can require extra work because PostgreSQL must identify duplicate result rows.

Do not automatically add `DISTINCT` when unexpected duplicates appear.

> **Important**
>
> Unexpected duplicates may signal a problem in the query logic, especially after joins. Understand why the duplicates exist before deciding to remove them.


# Part 11: COALESCE and Readable NULL Handling


## What COALESCE does

`COALESCE` returns the first non-`NULL` expression in its argument list.

```sql
COALESCE(value, fallback)
```

If `value` is present, PostgreSQL returns it. If it is `NULL`, PostgreSQL returns `fallback`.


## Text example

```sql
SELECT address,
       COALESCE(address2, 'No unit') AS unit
FROM address;
```

This does not change the stored data. It changes how the query result presents missing `address2` values.


## COALESCE and compatible types

The arguments to `COALESCE` must resolve to compatible data types.

For a timestamp displayed with a text fallback, cast the timestamp to text:

```sql
SELECT rental_id,
       customer_id,
       COALESCE(return_date::TEXT, 'Still out') AS return_status
FROM rental
ORDER BY rental_id;
```

`::TEXT` converts the timestamp value for this output expression.


## COALESCE in calculations

A `NULL` inside arithmetic can make the entire expression `NULL`.

A common analytical pattern is:

```sql
COALESCE(numeric_column, 0)
```

before arithmetic **when zero is a defensible business interpretation of missingness**.

> **Do not replace NULL blindly**
>
> `COALESCE` is a tool, not a rule. Replacing missing data with `0`, `'Unknown'`, or another fallback changes the interpretation of the output. Use a fallback only when it matches the analytical meaning.


## Try it: make missing output readable

```sql
SELECT address,
       address2,
       COALESCE(address2, 'No unit') AS unit
FROM address
ORDER BY address;
```

### Observe

Compare `address2` and `unit` for rows where `address2` is missing.

What changed in the result? What did **not** change in the underlying table?


# Part 12: Combining the Techniques


## One query, several ideas

```sql
SELECT DISTINCT
       district,
       address,
       COALESCE(address2, 'No unit') AS unit,
       postal_code || ', ' || district AS location_label
FROM address
ORDER BY district, address;
```

This query combines:

- `DISTINCT`
- selected columns
- `COALESCE`
- string concatenation
- aliases
- sorting


## Read a query in logical order

When a query becomes longer, do not read it only from top to bottom.

For the previous query, reason as follows:

```text
1. FROM address
2. no WHERE filter
3. no GROUP BY
4. no HAVING
5. SELECT expressions and aliases; DISTINCT shapes output
6. ORDER BY district, address
```

This habit becomes increasingly important as SQL grows more complex.

# Part 14: Debugging and AI Critique


## A query can look plausible and still be wrong

Consider this AI-generated SQL:

```sql
SELECT customer_id,
       COUNT(*) AS rental_count
FROM rental
WHERE COUNT(*) > 10
GROUP BY customer_id
ORDER BY rental_count DESC;
```

Do not begin by rewriting it. First identify the logical error.


## Critique the query

Answer these questions:

1. Which clause is misused?
2. What expression is being evaluated too early?
3. What exists when `WHERE` runs?
4. What does not exist yet?
5. Which clause should filter the grouped result?


## Corrected version

```sql
SELECT customer_id,
       COUNT(*) AS rental_count
FROM rental
GROUP BY customer_id
HAVING COUNT(*) > 10
ORDER BY rental_count DESC;
```

The important skill is not memorizing this correction. It is being able to **justify it from execution order**.


## A second AI error

```sql
SELECT title,
       rental_rate * 7 AS weekly_cost
FROM film
WHERE weekly_cost < 10.00
ORDER BY weekly_cost;
```

### Diagnose before fixing

- `weekly_cost` is created in which clause?
- `WHERE` runs before or after that clause?
- Why can `ORDER BY` use the alias even though `WHERE` cannot?


## Correct the second error

One valid correction is:

```sql
SELECT title,
       rental_rate * 7 AS weekly_cost
FROM film
WHERE rental_rate * 7 < 10.00
ORDER BY weekly_cost;
```

You could also reason from the original stored value if an equivalent simpler condition exists.

## Lab habit: translate before coding

For each question, write a short plan:

```text
OUTPUT: Which columns or expressions?
SOURCE: Which table?
ROWS: Which conditions?
ORDER: How should results be sorted?
LIMIT: Is a row cap required?
NULLS: How should missing values behave?
```

Then translate the plan into SQL.


## Lab habit: verify the result

After a query runs, ask:

- Does the row count seem plausible?
- Do the boundary values satisfy the condition?
- Are any unexpected `NULL` values present?
- Does the sort order match the question?
- Did parentheses preserve the intended logic?
- Did `DISTINCT` remove the kind of duplication I intended?


# Part 17: Check Your Understanding


## Knowledge check 1

Why does this fail?

```sql
SELECT title, length / 60.0 AS hours
FROM film
WHERE hours > 2;
```

### Answer

`WHERE` logically runs before `SELECT`, so the alias `hours` does not yet exist when the row filter is evaluated.


## Knowledge check 2

Which condition includes both endpoints 75 and 105?

A. `length > 75 AND length < 105`  
B. `length BETWEEN 75 AND 105`  
C. `length IN (75, 105)`

### Answer

**B.** `BETWEEN` is inclusive at both ends.


## Knowledge check 3

What does `%` mean in a `LIKE` pattern?

### Answer

Zero or more characters.

What does `_` mean?

Exactly one character.


## Knowledge check 4

Which is the correct missing-value test?

A. `return_date = NULL`  
B. `return_date IS NULL`

### Answer

**B.** Use `IS NULL` or `IS NOT NULL` for missing values.


## Knowledge check 5

What does this return?

```sql
SELECT DISTINCT rating, rental_rate
FROM film;
```

### Answer

One row for each unique **combination** of `rating` and `rental_rate` in the result.


## Knowledge check 6

Why can this work?

```sql
SELECT amount * 1.10 AS total
FROM payment
ORDER BY total DESC;
```

### Answer

`ORDER BY` logically runs after `SELECT`, so the alias `total` has already been defined.


## Knowledge check 7

Suppose `address2` is `NULL`. What does this expression return?

```sql
COALESCE(address2, 'No unit')
```

### Answer

`'No unit'`, because `COALESCE` returns the first non-`NULL` argument.


# Part 18: Wrap-up


## Module 2 concept map

```text
SELECT QUERY
│
├── Logical order
│   ├── FROM
│   ├── WHERE
│   ├── GROUP BY
│   ├── HAVING
│   ├── SELECT
│   ├── ORDER BY
│   └── LIMIT
│
├── Filter rows
│   ├── = <> < > <= >=
│   ├── AND / OR / NOT
│   ├── BETWEEN
│   ├── IN / NOT IN
│   ├── LIKE / ILIKE
│   └── IS NULL / IS NOT NULL
│
└── Shape output
    ├── expressions
    ├── aliases
    ├── DISTINCT
    └── COALESCE
```


## Module 2 takeaways

1. SQL's writing order and logical execution order are different.
2. `WHERE` filters rows before `SELECT` aliases exist.
3. `HAVING` is used for conditions on grouped results.
4. Parentheses are essential when mixed `AND`/`OR` logic could be ambiguous.
5. `BETWEEN` is inclusive, but timestamp boundaries require care.
6. `IN` makes list membership clearer than repeated `OR` conditions.
7. `LIKE` uses `%` for any sequence and `_` for exactly one character; PostgreSQL's `ILIKE` is case-insensitive.
8. `NULL` requires `IS NULL` or `IS NOT NULL`.
9. `DISTINCT` applies to the entire selected row combination.
10. `COALESCE` can make missing values readable, but the fallback must make analytical sense.


## Before you move on

You are ready for the Module 2 assessments when you can do all of the following without copying a finished query:

- write a multi-condition `WHERE` clause
- explain a difference caused by `AND`/`OR` precedence
- use `BETWEEN`, `IN`, `LIKE`, and `IS NULL`
- create and sort by a `SELECT` alias
- use `DISTINCT` intentionally
- use `COALESCE` with an appropriate fallback
- diagnose an alias-in-`WHERE` error
- diagnose an aggregate-in-`WHERE` error
- explain the correction using logical execution order
