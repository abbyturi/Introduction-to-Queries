# Part 1: Describing the World with Data


## What is data?

A useful starting point from Chapter 1 is that **data records descriptions or measurements of something in the real world**.

A database does not store the real customer, film, payment, or rental.

It stores selected facts about those things.

### Activity: what gets recorded and what does not?

Choose **one** setting: a university, hospital, retailer, or employer.

- Identify one real-world object or event the organization might record.
- List three facts it might store about that object or event.
- Identify two facts it might leave out.
- Explain why the included facts might be useful for a particular question.

**Key idea:** A dataset is a purposeful selection of facts about the real world; it is not the real world itself.


## Four terms to keep separate

| Term | Meaning | Example |
|---|---|---|
| Unit of observation | The object or event being described | one film rental |
| Variable | A characteristic measured or recorded | rental date |
| Data point | Recorded measurements for one observation | one rental record |
| Dataset | A collection of data points | all rental records |

These distinctions help us decide what a row should represent.


## Quick check

Suppose a university stores:

| student_id | course | term | final_grade |
|---:|---|---|---:|
| 10427 | ALY6420 | Fall 2026 | 91 |

**Think, then reflect:**

1. What is the unit of observation?
2. Which fields are variables?
3. What does this row not tell us?
4. What question could this data help answer?


## Relevance comes before volume

A system can record many facts, but not every fact is useful for every analytical question.

> **Analytical habit**
>
> Start with the question. Then identify the observations and variables needed to answer it.


For academic performance, course history and grades may be relevant. Shoe size probably is not.


## Three broad forms of data

### Structured
A defined organization and schema.

Examples:

- relational tables
- payroll records
- transaction data
### Semi-structured
Organized, but with a more flexible structure.

Examples:

- JSON
- XML
- API responses
### Unstructured
No fixed tabular schema.

Examples:

- document text
- images
- audio
- email bodies

# Part 2: From Data to a Model


## Why model data?

Chapter 1 defines data modeling as identifying relevant data and its relationships.

A model helps answer questions such as:

- What kinds of things exist in the system?
- What should we record about each thing?
- What uniquely identifies it?
- How is it related to other things?
- How should those ideas become database structures?


## The three modeling levels

```text
CONCEPTUAL
What exists and how is it related?
        |
        v
LOGICAL
What attributes and keys describe it?
        |
        v
PHYSICAL
How will the database implement it?
```

Each level adds detail.


## Conceptual model

A conceptual model describes the important real-world concepts without committing to database implementation details.

For a film rental business, we might identify:

- Customer
- Film
- Rental
- Payment
- Store

At this stage, we care about meaning and relationships.


## Logical model

A logical model adds attributes, identifiers, and relationship structure.

For example:

```text
CUSTOMER
customer_id
first_name
last_name
email

RENTAL
rental_id
rental_date
customer_id
inventory_id
```

Now we can see which attributes belong to each entity.


## Physical model

A physical model translates the design into database objects.

That includes details such as:

- table names
- column names
- data types
- primary key constraints
- foreign key constraints
- indexes

PostgreSQL operates at this implementation level.


## Entity-relationship thinking

Before writing a query, ask:

- What entity or event does each row represent?
- Which attributes describe it?
- Which column identifies it?
- Which columns connect it to other tables?

These questions will become essential when we work with joins later in the course.


# Part 3: Relational Databases


## What is a relational database?

A relational database organizes structured data into **relations**, commonly presented as tables.

A table contains:

- rows representing records
- columns representing attributes
- defined data types
- constraints that can protect valid structure and relationships

SQL is the standard language used to interact with relational database systems.


## Rows and columns

| film_id | title | release_year | rental_rate |
|---:|---|---:|---:|
| 1 | Academy Dinosaur | 2006 | 0.99 |
| 2 | Ace Goldfinger | 2006 | 4.99 |

**Row:** one film record  
**Column:** one attribute recorded consistently across film records

The table structure gives the data a shared meaning.


## Data types are part of the structure

Columns do not simply have names. They also have types.

Common PostgreSQL type families include:

| Type family | Typical use |
|---|---|
| Integer | IDs, counts, whole numbers |
| Numeric | prices, rates, precise decimal values |
| Character | names, titles, codes |
| Boolean | true or false conditions |
| Date and time | dates, timestamps, durations |

> **Why types matter**
>
> A data type communicates what values are valid and what operations make sense for a column.


## Table, schema, database

These terms describe different levels of organization.

```text
Database
  |
  +-- Schema
        |
        +-- Table
              |
              +-- Columns
              +-- Rows
```

A **schema** is a named organizational container for database objects.

In PostgreSQL, `public` is a common default schema.


## Primary keys

A **primary key** uniquely identifies a row.

Examples:

- `customer_id`
- `film_id`
- `rental_id`

A useful key should be unique and stable.

> **Key check**
>
> For each possible identifier below, ask: **Is it unique? Is it stable? Could it change?**
>
> - student email address
> - employee ID
> - a combination of course code + term + student ID
>
> Use your answers to predict which might function as a natural, compound, or surrogate key.


## Natural, compound, and surrogate keys

**Natural key**  
Uses an existing real-world attribute that is already unique.

**Compound key**  
Uses more than one column together to identify a row.

**Surrogate key**  
Uses an artificial identifier created for the database, often an integer ID.

Why might a stable surrogate ID be preferable to a person's email address?


## Foreign keys

A **foreign key** is a column that refers to a key in another table.

```text
CUSTOMER                         RENTAL
--------                         ------
customer_id  PK  <------------- customer_id  FK
first_name                       rental_id    PK
last_name                        rental_date
```

The foreign key records the relationship without repeating all customer details in every rental row.


## Referential integrity

A valid relationship should point to a real row.

If a rental contains `customer_id = 42`, a corresponding customer should exist.

Foreign key constraints can help the database enforce that rule.

> **Analytical value**
>
> Reliable relationships matter because later queries will combine information across tables. Broken references create misleading or incomplete results.


## Why normalization appears in Chapter 1

Normalization is introduced as a way to organize data so that unnecessary repetition and update problems are reduced.

Consider storing this on every rental:

```text
customer_id | customer_name | customer_email | rental_id | rental_date
```

If a customer's email changes, how many rows might need to be updated?


## A more relational design

Instead, separate customer facts from rental facts.

```text
CUSTOMER
customer_id | customer_name | customer_email

RENTAL
rental_id | customer_id | rental_date
```

The relationship is preserved through `customer_id`.

> **Module 1 level**
>
> Focus on the purpose of normalization. Detailed normal forms are not required for today's first queries.


## Relational databases: strengths

Relational systems are useful when we need:

- clear, consistent structure
- relationships among datasets
- integrity rules and constraints
- flexible querying
- transactional reliability
- a widely used standard query language


## Relational databases: tradeoffs

No data model is ideal for every problem.

Relational systems can require:

- deliberate schema design
- management of relationships across tables
- schema changes when requirements evolve
- additional design choices for highly irregular or nested data

The key question is whether the relational model fits the problem.


# Part 4: PostgreSQL and the Course Environment


## What is a DBMS?

A **database management system**, or DBMS, is software that manages databases and processes requests against them.

PostgreSQL is the relational DBMS used in this course.

SQL is the language we use to express requests to PostgreSQL.

```text
SQL statement -> PostgreSQL -> database objects -> result set
```


## PostgreSQL architecture, simplified

A PostgreSQL environment separates the client from the database server.

```text
Client application
      |
      | SQL request
      v
PostgreSQL server
      |
      | reads and processes
      v
Database
      |
      v
Result set returned to client
```

This client-server idea is important even when the server is hosted in the cloud.


## Our course tools

### PostgreSQL
The relational database management system that stores and processes course data.
### Neon
The cloud service hosting the PostgreSQL environment used for course work.
### DBeaver
The desktop client used to connect, browse database objects, write SQL, and inspect results.


## Do not confuse the layers

```text
You
 |
 v
DBeaver
 |
 | database connection
 v
Neon
 |
 v
PostgreSQL
 |
 v
Pagila database objects
```

**DBeaver is the client.**  
**Neon provides cloud hosting.**  
**PostgreSQL is the DBMS.**  
**Pagila is the course practice dataset.**

# Part 5: SQL as a Query Language


## What does SQL do?

SQL is declarative.

You describe the result you want. PostgreSQL determines how to produce it.

```sql
SELECT title, rating
FROM film
WHERE rating = 'G';
```

The query specifies the output columns, source table, and condition.


## The first query pattern

```sql
SELECT column_1, column_2
FROM table_name;
```

Read it as:

> Return these columns from this table.

This simple pattern is the foundation for much of the SQL you will write.


## `SELECT *`

```sql
SELECT *
FROM film;
```

`*` requests all columns available from the table.

It is useful for initial exploration, but it is usually better to request only the columns you need.


## Select specific columns

```sql
SELECT film_id, title, rating, rental_rate
FROM film;
```

Benefits of being explicit:

- the result is easier to read
- the intent is clearer
- fewer unnecessary columns are transferred
- downstream work is less dependent on the table's full structure


## Query result sets

A `SELECT` query produces a **result set**.

The result set is not the same thing as the source table.

```text
Source table
   |
   | SELECT columns
   | filter rows
   | sort output
   v
Result set
```

A query can present a useful view of the stored data without changing the stored rows.


## Filter rows with `WHERE`

```sql
SELECT title, rating
FROM film
WHERE rating = 'G';
```

`WHERE` keeps rows that satisfy a condition.

Text values use single quotes.


## Comparison operators

Common comparisons include:

| Operator | Meaning |
|---|---|
| `=` | equal to |
| `<>` or `!=` | not equal to |
| `<` | less than |
| `>` | greater than |
| `<=` | less than or equal to |
| `>=` | greater than or equal to |


## Numeric filtering

```sql
SELECT title, rental_rate
FROM film
WHERE rental_rate < 2.00;
```

The condition is evaluated for each candidate row.

Ask yourself:

> What must be true for a row to remain in the result?


## Combine simple conditions

```sql
SELECT title, rating, rental_rate
FROM film
WHERE rating = 'G'
  AND rental_rate < 2.00;
```

With `AND`, both conditions must be true.

```sql
SELECT title, rating
FROM film
WHERE rating = 'G'
   OR rating = 'PG';
```

With `OR`, either condition may be true.

> **Module 1 boundary**
>
> Practice simple comparisons with `WHERE` and combine them with `AND` or `OR`. Do **not** use `IN`, `BETWEEN`, or `NULL`-handling techniques yet; those are covered in Module 2.


## Sort with `ORDER BY`

A relational query does not guarantee display order unless you request one.

```sql
SELECT title, rental_rate
FROM film
ORDER BY rental_rate DESC;
```

- `ASC` sorts ascending
- `DESC` sorts descending
- ascending is the default when no direction is written


## Sort by more than one column

```sql
SELECT title, rating, rental_rate
FROM film
ORDER BY rating ASC, title ASC;
```

PostgreSQL sorts first by `rating` and then by `title` within equal rating values.

This is useful when one sort column contains repeated values.


## Limit the displayed rows

```sql
SELECT title, rating, rental_rate
FROM film
LIMIT 10;
```

`LIMIT` is especially useful while exploring an unfamiliar table.

> **Important distinction**
>
> `LIMIT` reduces the rows returned to the client. Do not treat it as a general server-side performance optimization strategy.


## Build a query in layers

```sql
SELECT title, rating, rental_rate
FROM film
WHERE rating = 'G'
ORDER BY rental_rate DESC
LIMIT 5;
```

Read it in plain language:

> From `film`, keep G-rated films, return the selected columns, order the result by rental rate from highest to lowest, and display five rows.


## Clause roles

| Clause | Question it answers |
|---|---|
| `SELECT` | Which columns or expressions should appear? |
| `FROM` | Which table supplies the data? |
| `WHERE` | Which rows qualify? |
| `ORDER BY` | How should the result be displayed? |
| `LIMIT` | How many rows should be returned? |

Do not memorize only the order. Understand the role of each clause.


## Predict before you run

What will this query return?

```sql
SELECT title, rating
FROM film
WHERE rating = 'PG'
ORDER BY title
LIMIT 3;
```

Identify:

1. the source table
2. the output columns
3. the row condition
4. the sort
5. the maximum result size


## Debug by reading the error

Suppose this fails:

```sql
SELECT title, rating
FROM films
WHERE rating = "G";
```

What should you check?

- Is the table name correct?
- Is the text value quoted correctly?
- Does the named column exist?
- Are you connected to the intended database?


## Common Module 1 mistakes

- confusing the client application with the database server
- querying the wrong connection or database
- misspelling a table or column name
- using double quotes for ordinary text values
- assuming rows are sorted without `ORDER BY`
- selecting every column when only a few are needed
- writing a condition without first deciding what the row represents
