# 1. Understanding the Tools

Before installing anything, it is helpful to understand what each tool does.

### PostgreSQL: The Database System

**PostgreSQL** is the relational database management system (RDBMS).

It stores information in tables consisting of **rows and columns** and processes commands written in **SQL (Structured Query Language).**

For example:

```sql
SELECT *
FROM employees
WHERE department = 'Finance';
```

**Think of PostgreSQL as the database engine** that stores, organizes, and processes our data.

---

### Neon: The Cloud Host

PostgreSQL must run on a computer or server.

Instead of installing and maintaining a PostgreSQL server on your personal computer, we will use **Neon**, which hosts PostgreSQL in the cloud.

Neon provides the information DBeaver needs to connect to your database, including:

* Host/server
* Database name
* Username
* Password
* Port
* SSL/security settings

**Think of Neon as the cloud home where our PostgreSQL database lives.**

---

### DBeaver: The Database Workspace

**DBeaver Community Edition** is the desktop application we will use to interact with PostgreSQL.

With DBeaver, you can:

* Connect to PostgreSQL
* View databases, schemas, and tables
* Write SQL queries
* Execute SQL
* View query results
* Create and modify database objects
* Import database scripts

**Think of DBeaver as your workspace for communicating with PostgreSQL.**

---

### Pagila: The Practice Database

**Pagila** is a sample PostgreSQL database containing related tables and realistic sample data.

We will use Pagila throughout the course to practice SQL.

---

# Troubleshooting Guide

The following are some common issues you may encounter during setup.

---

## Problem 1: `function uuidv7() does not exist`

### Error

```text
ERROR: function uuidv7() does not exist
```

### Likely Cause

You are using the **current Pagila files with an older PostgreSQL version**, such as PostgreSQL 16.

### Solution

Run:

```sql
SELECT version();
```

Verify that your Neon database is running:

```text
PostgreSQL 18
```

If you created your Neon project using PostgreSQL 16, create/use a PostgreSQL 18 project for the current course setup.

---

# Problem 2: `role "postgres" does not exist`

You may encounter:

```text
SQL Error [42704]:
ERROR: role "postgres" does not exist
```

This does **not** necessarily mean PostgreSQL or Neon is broken.

The SQL dump may contain commands such as:

```sql
ALTER TABLE public.store OWNER TO postgres;
```

The database from which the dump was originally created had a database role/user named:

```text
postgres
```

Your Neon database may instead use:

```text
neondb_owner
```

The script is therefore essentially asking Neon:

```text
Create this object
      ↓
Assign its ownership to "postgres"
```

Neon responds:

```text
I do not have a role named "postgres."
```

## Confirm Your Actual User

Run:

```sql
SELECT current_user;
```

If the result is:

```text
neondb_owner
```

that is your Neon database user.

## Fix the Ownership References

In DBeaver, open **Find and Replace**:

```text
Ctrl + H
```

Find:

```text
OWNER TO postgres
```

Replace with your actual Neon username, for example:

```text
OWNER TO neondb_owner
```

Then select **Replace All**.

For example:

```sql
ALTER TABLE public.store OWNER TO postgres;
```

becomes:

```sql
ALTER TABLE public.store OWNER TO neondb_owner;
```

For a temporary class/sample database, ownership statements can sometimes simply be removed or commented out:

```sql
-- ALTER TABLE public.store OWNER TO postgres;
```

Also search for:

```text
TO postgres
```

because the dump may contain other executable commands such as:

```sql
GRANT ... TO postgres;
```

### Important

A line such as:

```sql
-- Owner: postgres
```

is only a **comment**.

It does not need to be changed.

The problem is executable SQL such as:

```sql
ALTER ... OWNER TO postgres;
```

---

# Problem 3: DBeaver Asks for `:4`, `:20`, `:16`, etc.

You may see a popup titled something similar to:

```text
Bind parameter(s)
```

containing entries such as:

```text
:4
:20
:16
:11
:14
:8
```

### Do NOT enter values for these.

DBeaver may be incorrectly interpreting text inside the SQL dump as SQL parameters.

Normally syntax such as:

```sql
WHERE employee_id = :employee_id
```

can represent a parameter.

However, data containing values resembling:

```text
12:20
8:16
14:11
```

may cause DBeaver to interpret the colon-number combinations as parameters.

### What to Do

Select:

**Ignore**

Do **not** spend time entering values into the parameter Value column.

---

# Problem 4: DBeaver Shows "Not Responding"

If DBeaver shows:

```text
Not Responding
```

or:

```text
Datasource was invalidated
```

or indicates that all available connections are being used, **do not repeatedly click Execute**.

A very large SQL script can consume significant resources while DBeaver tries to parse and execute it.

Give DBeaver some time to finish.

If the application remains frozen, close and reopen DBeaver.

After reopening it, reconnect to Neon and run:

```sql
SELECT 1;
```

If the result is:

```text
1
```

your Neon database connection is working.

---

# Problem 5: I/O Error While Sending to Backend

You may encounter:

```text
SQL Error [08006]:
An I/O error occurred while sending to the backend.
```

This often means the connection between DBeaver and the PostgreSQL server was interrupted while a large script was being transmitted or executed.

### What to Do

First reconnect and test:

```sql
SELECT 1;
```

If this succeeds, the database itself is reachable.

Avoid immediately rerunning the enormous script by selecting all of its contents again.

Instead, use DBeaver's appropriate **script execution / native PostgreSQL execution** mechanism for the import.

---

# Problem 6: "Relation Already Exists"

If a previous import stopped halfway through, some database objects may already exist.

A second attempt can therefore produce errors such as:

```text
relation already exists
```

If this is a **new class database and you have no work that needs to be preserved**, you can reset the `public` schema:

```sql
DROP SCHEMA public CASCADE;

CREATE SCHEMA public;
```

Then start the Pagila import again from the beginning:

```text
1. pagila-schema.sql
2. pagila-data.sql
```

> **WARNING:** `DROP SCHEMA public CASCADE;` deletes objects in the `public` schema. Do not use this command on a database containing work you need to keep.

---

# Problem 7: Duplicate-Key Errors

If Pagila was already successfully loaded and you execute `pagila-data.sql` again, you may receive duplicate-key or duplicate-record errors.

This does **not** necessarily mean the original import failed.

Before importing again, check whether the data is already there:

```sql
SELECT COUNT(*)
FROM actor;
```

and:

```sql
SELECT COUNT(*)
FROM film;
```

If the tables already contain the expected data, **do not reload `pagila-data.sql`.**

---

# Problem 8: Multiple DBeaver Connections

You may accidentally create connections named something like:

```text
postgres
postgres 2
```

This can become confusing because both may point to the same Neon database.

For this course, you generally need **only one working connection**.

A simple environment should look like:

```text
DBeaver
   ↓
PostgreSQL Connection
   ↓
Neon
   ↓
PostgreSQL 18
   ↓
neondb
   ↓
Pagila
```

Before executing SQL, check the active connection/database displayed in the SQL Editor.

Make sure you are working with your **ALY6420 Neon database**.

---

# Recommended Troubleshooting Order

When something goes wrong, **read the error message before changing settings**.

Use this sequence:

```text
Something failed
      ↓
Read the error message
      ↓
Run SELECT 1;
      ↓
Does it work?
   ↙        ↘
 YES        NO
  ↓          ↓
Connection   Reconnect /
is okay      check Neon
  ↓
Run SELECT version();
  ↓
Confirm PostgreSQL 18
  ↓
Run SELECT current_user;
  ↓
Confirm Neon user
  ↓
Identify the specific
script/import error
```

Do not repeatedly execute the same large script without first determining what caused the previous attempt to fail.
